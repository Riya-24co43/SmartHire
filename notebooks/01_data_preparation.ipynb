```python
# Find resume categories and count resumes

import os

resume_data_path = "../data/raw/Resume_dataset/data/data"

categories = []

for category in os.listdir(resume_data_path):
    category_path = os.path.join(resume_data_path, category)

    if os.path.isdir(category_path):
        pdf_files = [f for f in os.listdir(category_path) if f.lower().endswith(".pdf")]
        categories.append((category, len(pdf_files)))

print("Number of categories:", len(categories))

print("\nResume categories:")
for category, count in categories:
    print(category, ":", count, "resumes")
```

```python
# Extract text from all resume PDFs

import os
import PyPDF2
import pandas as pd

resume_data_path = "../data/raw/Resume_dataset/data/data"

resume_data = []

for category in os.listdir(resume_data_path):

    category_path = os.path.join(resume_data_path, category)

    if not os.path.isdir(category_path):
        continue

    for file in os.listdir(category_path):

        if file.lower().endswith(".pdf"):

            file_path = os.path.join(category_path, file)

            try:
                reader = PyPDF2.PdfReader(file_path)

                text = ""

                for page in reader.pages:
                    page_text = page.extract_text()

                    if page_text:
                        text += page_text + " "

                resume_data.append({
                    "Category": category,
                    "Resume_Text": text
                })

            except Exception as e:
                print("Error reading:", file)
                print(e)

print("Text extraction completed!")
print("Total resumes:", len(resume_data))
```

```python
# Convert extracted resumes into a DataFrame

resume_df = pd.DataFrame(resume_data)

print("Dataset shape:", resume_df.shape)

resume_df.head()
```

```python
# Check for empty resume text

empty_resumes = resume_df["Resume_Text"].str.strip().eq("").sum()

print("Empty resumes:", empty_resumes)
print("Total resumes:", len(resume_df))
```

```python
# Remove resumes with empty text

resume_df = resume_df[resume_df["Resume_Text"].str.strip() != ""]

print("Resumes after cleaning:", len(resume_df))
```

```python
# Check the number of resumes in each category

category_counts = resume_df["Category"].value_counts()

print(category_counts)
```

```python
# Save the cleaned resume dataset

os.makedirs("../data/processed", exist_ok=True)

resume_df.to_csv(
    "../data/processed/cleaned_resumes.csv",
    index=False
)

print("Cleaned resume dataset saved successfully!")
```

```python
X = resume_df["Resume_Text"]
y = resume_df["Category"]
print("Input data:", X.shape)
print("Target data:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training resumes:", len(X_train))
print("Testing resumes:", len(X_test))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    stop_words="english",
    max_features=5000
)

X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

print("TF-IDF training shape:", X_train_tfidf.shape)
print("TF-IDF testing shape:", X_test_tfidf.shape)
```

```python
from sklearn.linear_model import LogisticRegression

classifier = LogisticRegression(
    max_iter=1000
)

classifier.fit(X_train_tfidf, y_train)

print("Resume classifier trained successfully!")
```

```python
# Make predictions

y_pred = classifier.predict(X_test_tfidf)

print("Predictions completed!")
print("First 10 predictions:")
print(y_pred[:10])
```

```python
from sklearn.metrics import accuracy_score, classification_report

accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", round(accuracy * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
```
